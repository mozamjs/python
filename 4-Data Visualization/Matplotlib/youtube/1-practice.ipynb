```python
import matplotlib.pyplot as plt
# pyplot is a collection of functions that make matplotlib work as MATLAB
```

```python
# 1 create first chart
```

```python
# data 
x = [1, 2, 3, 4, 5]
y = [10,20,25,35,45]

plt.plot(x,y) # Line chart
plt.show()    # display chart
```

```python
# 2 Customize chart
```

```python
plt.figure(figsize =(6,4))

plt.plot(x,y, color='r', marker='o', linestyle='-.', linewidth=2, markersize =12 )
plt.title("Line chart")
plt.xlabel('Time')
plt.ylabel('speed')

plt.show()
```

```python
# 3 Advanced Line chart - multiple Lines
```

```python
x = [1, 2, 3, 4, 5]
y1 = [10,20,25,35,45]
y2 = [20,30,35,45,55]

plt.plot(x,y1, label = 'Sales 2024')
plt.plot(x,y2, label = 'Sales 2025')
plt.title ("yoy sales")
plt.xlabel('Month')
plt.ylabel('Sales')

plt.legend()
plt.show()
```

```python
# 4 Bar chart
```

```python
fruits = ['Mango', 'Oringe', 'Banana', 'Stwberry', 'Peach']
sales = [60, 30, 70, 90, 33]

plt.bar(fruits, sales )
plt.title("Bar Chart Example")
plt.xlabel("Fruites")
plt.ylabel('sales')
plt.show()
```

```python
#  5 Histogram
```

```python
# data  use for frequency distribution analysis
import random
data = [random.randint(1,10) for _ in range(100)]

plt.hist(data, bins = 5)
plt.title("Histogram Example")
plt.show()
```

```python
# 6 pie chart 
```

```python
# used to show part-to whole relationship/Data 
```

```python
fruits = ['Mango', 'Oringe', 'Banana', 'Stwberry', 'Peach']
sales = [60, 30, 70, 90, 33]

plt.pie(sales, labels = fruits, autopct = '%1.1f%%',startangle=90)
plt.title("Pie Chart Example")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.random.randint(100, size=(100))
y = np.random.randint(100, size=(100))
colors = np.random.randint(100, size=(100))
sizes = 10 * np.random.randint(100, size=(100))

plt.scatter(x, y, c=colors, s=sizes, alpha=0.5, cmap='nipy_spectral')

plt.colorbar()

plt.show()
```

```python
# 8 subplots
# Use to show multiple chart in one figure
```

```python
# data-1 bar chart

categproes = ['Mon','Tue','wed','Thu','Fri']
sales = [10,20,55,35,45]

# data-2 scatter plot
y1 = [10,20,30,40,50]
y2 = [35,20,11,30,16]

plt.figure(figsize=(8,4))

plt.subplot(1,2,1) # row, column, position 
plt.bar(categproes, sales)
plt.title("Daily Sales")
plt.xlabel("week days")
plt.ylabel("sales")





plt.subplot(1,2,2)
plt.scatter(y1,y2)
plt.title("user Example")
plt.xlabel("User1")
plt.ylabel("User2")

plt.show()


```

```python
# 9 matplotlib with pandas real data 
```

```python
# create df

import pandas as pd

# data 
data = {
    'Month' : ['Jan', 'Feb', 'Mar', 'Apr'],
    'Sales' : [12000, 11000, 13000, 25000]
}

df = pd.DataFrame(data)
df



```

```python
plt.bar(df['Month'],df['Sales'])
plt.title("Matplotlib with pandas")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
# save charts/plot
```

```python
plt.bar(df['Month'],df['Sales'])
plt.title("Matplotlib with pandas")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.savefig("Monthly_Sales_from_df.png")

plt.show()
```
