```python
# Basic line plot 

import matplotlib.pyplot as plt 

x = [1,2,3,4,5]
y = [18, 25, 21, 30, 25]

plt.plot(x,y)
plt.show()
```

```python
# Title & Labels

plt.plot(x,y)

plt.title("Student Marks")
plt.xlabel("students")
plt.ylabel("Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt

product = ['Mobile',"Laptop", "Tablet", "watch"]
sales = [80, 50, 35, 60]

plt.bar(product, sales)

plt.title("Product Sales")
plt.xlabel("products")
plt.ylabel("Sales")
```

```python
# exercise 

import matplotlib.pyplot as plt

Reviews = ["Positive", "Negative", "Neutral"]
Sentiment = [80, 45, 25]


plt.bar(Reviews, Sentiment)

plt.title("Customer Feedback")
plt.xlabel('Sentiment')
plt.ylabel('Number of Reviews')

plt.show()
```

```python
# Pie Chart

import matplotlib.pyplot as plt

sentiments = ['Positive', 'Negative', 'Neutral']

reviews = [60,25,15]

plt.pie(reviews, labels=sentiments, autopct = '%1.1f%%' )

plt.title("Customer Feedback")
plt.show()
```

```python
# Histogram 

import matplotlib.pyplot as plt

marks = [45, 52, 67, 71, 73, 80, 82, 65, 90,
         55, 60, 72, 88, 91, 76, 69, 58, 84, 95, 63]

plt.hist (marks, bins = 5)

plt.title("Student Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()


```

```python
# Scatter plot 

# Scatter plot ka main purpose hai do numerical variables ke beech relationship dekhna.

# Example:

# Kya student jitne zyada hours study karta hai, uske marks bhi zyada hote hain?
```

```python
import matplotlib.pyplot as plt

study_hours = [1,2,3,4,5,6,7]
marks = [35,42,50,58,65,75,85]

plt.scatter(study_hours, marks)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt

temperature = [20, 22, 25, 28, 30, 32, 35]
ice_cream_sales = [30, 35, 45, 55, 65, 75, 90]

plt.scatter(temperature, ice_cream_sales)

plt.title("Temperature vs Ice Cream Sales")
plt.xlabel("Temperature")
plt.ylabel("Ice Cream Sales")

plt.show()
```
