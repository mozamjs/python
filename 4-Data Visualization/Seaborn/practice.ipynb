```python

# Countplot()

import matplotlib.pyplot as plt
import seaborn as sns

sentiment = [
    "Positive",
    "Negative",
    "Positive",
    "Positive",
    "Negative",
    "Neutral",
    "Positive",
    "Negative",
    "Positive",
    "Neutral",
]

sns.countplot(x = sentiment)

plt.title("Customer Sentiment")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")

plt.show()
```

```python
import matplotlib.pyplot as plt 
import seaborn as sns

ages = [18, 19, 20, 20, 21, 21, 22, 22, 23, 24, 24, 25, 26, 27, 28, 30, 31, 32]

sns.histplot(ages, bins = 5)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of People")

plt.show() 
```

```python
# barplot() ka use generally different categories ke numerical values ko compare karne ke liye hota hai
```

```python
import matplotlib.pyplot as plt
import seaborn as sns 

products = ["sleepers", "sneeker", "scatchers", "jogures"]
sales = [52,85,46,79]

sns.barplot(x = products, y = sales)

plt.title("shop sales chart")
plt.xlabel("Products Name")
plt.ylabel("Sales Fig")

plt.show()


```

```python
# scatterplot()
# jab humein do numerical variables ka beech relationship dehkna ho 

import matplotlib.pyplot as plt 
import seaborn as sns

temperature = [20,22,25,28,30,32,35]
sales = [30,35,45,55,65,75,90]

sns.scatterplot(x = temperature, y = sales )

plt.title("Temperature vs Ice Cream Sales")
plt.xlabel("Temperature")
plt.ylabel("Ice Cream Sales")

plt.show()
```

```python
# boxplot() numerical data ki distribution ko summarize karta hai aur especially ye cheezein dekhne mein useful hai:

# Data ka center kahan hai
# Data kitna spread hai
# Koi unusual value (outlier) to nahi
```

```python
import matplotlib.pyplot as plt 
import seaborn as sns

marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 70, 72, 75, 78, 80, 95]

sns.boxplot(x = marks)

plt.title("Student Marks Distribution")
plt.xlabel("Marks")

plt.show()
```
