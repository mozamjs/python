```python
# Text normalization

text = "I REALLY LOVE This Product!"

print("Original text: ")
print(text)

text = text.lower()
print("\nLowercase text: ")
print(text)
```

```python

text = "I bought 5 phones in 2025."

print(text)
```

```python
# Number handling 
import re

text = "I bought 5 phones in 2025."

text = re.sub(r'\d+', '', text)

print(text)
```

```python
import re 

text = "This product has 5 stars and  I bought it in 2025."

clean_text = re.sub(r'\d+','',text)

print("Original:")
print(text)

print("\nAfter removing numbers:")   
print(clean_text)

```

```python
# whitespace & Punctuation Removal 

text = " I really love this product!!!  "
print("Orignal text")
print(text)

text = text.strip()
print(text)
```

```python
import string 

text = text.translate(str.maketrans('', '', string.punctuation))

# Yahan string.punctuation Python ki common punctuation characters ki list deta hai:

print(text)


```

```python
# Tokenization

text = "I really love this product"

tokens = text.split()

print(tokens)
```

```python
review = "I really enjoy using this phone"

tokens = review.split()

print(tokens)
```
